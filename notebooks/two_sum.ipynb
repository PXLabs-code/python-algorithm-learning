# 两数之和（Two Sum）

[![在 Google Colab 中打开](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/PXLabs-code/python-algorithm-learning/blob/main/notebooks/two_sum.ipynb)

## 题目

给定一个整数数组 `nums` 和一个整数目标值 `target`，请在数组中找出和为目标值的两个整数，并返回它们的下标。

- 每种输入只对应一个答案。
- 同一个元素不能重复使用。
- 返回的下标顺序不限。

**示例**：输入 `nums = [2, 7, 11, 15]`、`target = 9`，输出 `(0, 1)`。

## 思路

遍历数组时，用哈希表保存已经见过的数字及其下标。对当前数字 `value`，查找 `target - value` 是否已经出现。这样只需遍历一次。

- 时间复杂度：$O(n)$
- 空间复杂度：$O(n)$

```python
def two_sum(nums: list[int], target: int) -> tuple[int, int]:
    """返回两个不同元素的下标，使对应元素之和等于 target。"""
    seen: dict[int, int] = {}

    for index, value in enumerate(nums):
        complement = target - value
        if complement in seen:
            return seen[complement], index
        seen[value] = index

    raise ValueError("No two sum solution")
```

## 正确性断言与边界测试

覆盖常规输入、重复值、负数、零，以及不存在答案的情况。

```python
assert two_sum([2, 7, 11, 15], 9) == (0, 1)
assert two_sum([3, 2, 4], 6) == (1, 2)
assert two_sum([3, 3], 6) == (0, 1)
assert two_sum([-3, 4, 3, 90], 0) == (0, 2)
assert two_sum([0, 4, 3, 0], 0) == (0, 3)

for nums, target in [([], 0), ([1], 1), ([1, 2], 4)]:
    try:
        two_sum(nums, target)
    except ValueError:
        pass
    else:
        raise AssertionError(f"应当找不到答案：{nums=}, {target=}")

print("所有断言均已通过。")
```

## `timeit` 效率测试

使用多次重复测试并显示最佳耗时，降低单次测量波动的影响。

```python
from timeit import repeat

benchmark_nums = list(range(10_000))
benchmark_target = 19_997
runs = repeat(
    lambda: two_sum(benchmark_nums, benchmark_target),
    number=100,
    repeat=5,
)
print(f"每次调用最佳耗时：{min(runs) / 100 * 1_000:.3f} ms")
```
